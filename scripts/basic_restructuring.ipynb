```python
import pandas as pd

file_path = r"D:\Mine\data_set\online_retail_II.xlsx"

# Read both sheets
df1 = pd.read_excel(file_path, sheet_name="Year 2009-2010")
df2 = pd.read_excel(file_path, sheet_name="Year 2010-2011")

# Combine them
df = pd.concat([df1, df2], ignore_index=True)

# Save as CSV
df.to_csv(
    r"D:\Mine\data_set\online_retail.csv",
    index=False,
    encoding="utf-8"
)

print(df.shape)
```
